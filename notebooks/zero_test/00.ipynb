```python
import torch
from millify import millify
from eshmun.models.zero.models import EshmunZeroModel, EshmunForMaskedLM
from eshmun.models.zero.configuration import EshmunZeroConfig
```

```python
config = EshmunZeroConfig(hidden_size=384, num_hidden_layers=8, intermediate_size=512, num_attention_heads=8)
```

```python
config
```

```python
model = EshmunForMaskedLM(config)

model
```

```python
input_ids = torch.randint(0, config.vocab_size, (2, 128))
labels = input_ids.clone()

out = model(input_ids=input_ids, labels=labels)

out.loss
```

```python
out.loss.exp()
```

```python
num_params = sum([p.nelement() for p in model.eshmun.encoder.parameters()])

print(millify(num_params), num_params)
```
