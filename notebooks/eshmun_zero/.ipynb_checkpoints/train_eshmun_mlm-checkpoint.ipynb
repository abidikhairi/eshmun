# Training Eshmun-Zero for Masked Language Modeling

This notebook demonstrates how to train an `EshmunForMaskedLM` model from scratch using the Hugging Face `Trainer` API.

**Pipeline:**
1. Build/load a tokenizer
2. Instantiate the config and model
3. Load and tokenize a dataset
4. Set up the MLM data collator
5. Train with `Trainer`
6. Run a quick fill-mask sanity check

## 1. Imports

```python
import torch
from datasets import load_dataset
from transformers import (
    PreTrainedTokenizerFast,
    DataCollatorForLanguageModeling,
    Trainer,
    TrainingArguments,
)

from eshmun.models.zero.config import EshmunZeroConfig
from eshmun.models.zero.models import EshmunForMaskedLM

print("CUDA available:", torch.cuda.is_available())
```

## 2. Tokenizer

The default `vocab_size=25` in the config suggests a character/protein-style alphabet rather than BPE. Replace this with your own tokenizer (e.g. one trained on Phoenician characters). For the sake of a runnable example, we build a tiny character-level tokenizer here.

```python
from tokenizers import Tokenizer, models, pre_tokenizers, trainers, processors

# Special tokens
SPECIAL_TOKENS = ["[PAD]", "[UNK]", "[CLS]", "[SEP]", "[MASK]"]

# A character-level tokenizer is a sensible default given vocab_size=25
tok = Tokenizer(models.WordLevel(unk_token="[UNK]"))
tok.pre_tokenizer = pre_tokenizers.Split(pattern="", behavior="isolated")

trainer = trainers.WordLevelTrainer(
    vocab_size=25,
    special_tokens=SPECIAL_TOKENS,
)

# Train on a small corpus — replace with your real data iterator
sample_corpus = [
    "abcdefghijklmnopqrstuvwxyz",
    "the quick brown fox jumps over the lazy dog",
]
tok.train_from_iterator(sample_corpus, trainer=trainer)

tok.post_processor = processors.TemplateProcessing(
    single="[CLS] $A [SEP]",
    pair="[CLS] $A [SEP] $B:1 [SEP]:1",
    special_tokens=[
        ("[CLS]", tok.token_to_id("[CLS]")),
        ("[SEP]", tok.token_to_id("[SEP]")),
    ],
)

tokenizer = PreTrainedTokenizerFast(
    tokenizer_object=tok,
    pad_token="[PAD]",
    unk_token="[UNK]",
    cls_token="[CLS]",
    sep_token="[SEP]",
    mask_token="[MASK]",
    model_max_length=512,
)

print("Vocab size:", tokenizer.vocab_size)
print("Sample:", tokenizer("hello").tokens())
```

## 3. Config & Model

Use a small config so the notebook runs quickly. Scale up `hidden_size`, `num_hidden_layers`, etc. for real training.

```python
config = EshmunZeroConfig(
    vocab_size=tokenizer.vocab_size,
    hidden_size=128,
    num_hidden_layers=4,
    num_attention_heads=4,
    intermediate_size=512,
    max_position_embeddings=514,
    local_window_size=12,
    alpha_init=0.0,
    pad_token_id=tokenizer.pad_token_id,
    is_decoder=False,
)

model = EshmunForMaskedLM(config)

n_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
print(f"Trainable parameters: {n_params/1e6:.2f}M")
```

## 4. Dataset

Replace the dataset below with your own (e.g. a corpus of Phoenician inscriptions, protein sequences, etc.). Here we use `wikitext-2` purely as a runnable placeholder.

```python
raw_datasets = load_dataset("wikitext", "wikitext-2-raw-v1")
raw_datasets = raw_datasets.filter(lambda ex: len(ex["text"].strip()) > 0)
print(raw_datasets)
```

```python
MAX_LEN = 128

def tokenize_fn(examples):
    return tokenizer(
        examples["text"],
        truncation=True,
        max_length=MAX_LEN,
        return_special_tokens_mask=True,
    )

tokenized = raw_datasets.map(
    tokenize_fn,
    batched=True,
    remove_columns=["text"],
    desc="Tokenizing",
)
tokenized
```

### (Optional) Group texts into fixed-length blocks

MLM training is more efficient on packed, fixed-length sequences. Skip this if your examples are already roughly uniform in length.

```python
BLOCK_SIZE = 128

def group_texts(examples):
    concatenated = {k: sum(examples[k], []) for k in examples.keys()}
    total_len = (len(concatenated["input_ids"]) // BLOCK_SIZE) * BLOCK_SIZE
    return {
        k: [t[i : i + BLOCK_SIZE] for i in range(0, total_len, BLOCK_SIZE)]
        for k, t in concatenated.items()
    }

lm_datasets = tokenized.map(group_texts, batched=True, desc="Grouping into blocks")
lm_datasets
```

## 5. Data Collator

`DataCollatorForLanguageModeling` handles the BERT-style 80/10/10 masking strategy and creates `labels` with `-100` at non-masked positions.

```python
data_collator = DataCollatorForLanguageModeling(
    tokenizer=tokenizer,
    mlm=True,
    mlm_probability=0.15,
)
```

## 6. Training

Adjust `num_train_epochs`, batch size, and learning rate to taste. The numbers below are minimal so the notebook runs quickly.

```python
training_args = TrainingArguments(
    output_dir="./eshmun-zero-mlm",
    overwrite_output_dir=True,
    num_train_epochs=1,
    per_device_train_batch_size=32,
    per_device_eval_batch_size=32,
    learning_rate=5e-4,
    weight_decay=0.01,
    warmup_ratio=0.06,
    lr_scheduler_type="linear",
    logging_steps=50,
    eval_strategy="epoch",
    save_strategy="epoch",
    save_total_limit=2,
    fp16=torch.cuda.is_available(),
    report_to="none",
)

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=lm_datasets["train"],
    eval_dataset=lm_datasets["validation"],
    data_collator=data_collator,
    tokenizer=tokenizer,
)
```

```python
train_result = trainer.train()
trainer.save_model()
tokenizer.save_pretrained(training_args.output_dir)
train_result.metrics
```

## 7. Evaluation

```python
import math

eval_metrics = trainer.evaluate()
eval_metrics["perplexity"] = math.exp(eval_metrics["eval_loss"])
eval_metrics
```

## 8. Sanity check: fill a mask

Quick qualitative test that the model produces something other than uniform noise.

```python
model.eval()
device = next(model.parameters()).device

text = f"the quick brown fox {tokenizer.mask_token} over the lazy dog"
inputs = tokenizer(text, return_tensors="pt").to(device)

with torch.no_grad():
    logits = model(**inputs).logits

mask_idx = (inputs.input_ids == tokenizer.mask_token_id).nonzero(as_tuple=True)[1]
top_ids = logits[0, mask_idx].topk(5, dim=-1).indices[0]
print("Top-5 predictions:", tokenizer.convert_ids_to_tokens(top_ids))
```

## 9. (Optional) Inspect alpha gates

Since Eshmun-Zero exposes a learnable per-layer gate between local and global attention, it's useful to inspect what the model learned.

```python
with torch.no_grad():
    out = model(**inputs, output_alphas=True)

if getattr(out, "alphas", None) is not None:
    for i, a in enumerate(out.alphas):
        print(f"Layer {i}: alpha = {a.mean().item():.3f}")
else:
    print("Model did not return alphas — check that EshmunZeroModel forwards output_alphas.")
```
