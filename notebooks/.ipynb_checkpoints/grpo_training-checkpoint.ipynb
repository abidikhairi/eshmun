**Pipeline:**
1. Install Eshmun from GitHub
2. Login to Weights & Biases
3. Load model, tokenizer and dataset from HuggingFace Hub
4. Configure reward functions and GRPO hyperparameters
5. Train with W&B logging
6. Save the fine-tuned model

## 3. Imports

```python
import logging
import re

import torch
from datasets import load_dataset
from transformers import AutoModelForCausalLM, AutoTokenizer

import eshmun.trainer.grpo.reward_functions as _rf  # registers all reward functions
from eshmun.trainer.grpo.config import GRPOConfig
from eshmun.trainer.grpo.trainer import GRPOTrainer
from eshmun.trainer.grpo.reward_functions.protein import (
    AminoAcidRepetitionPenalty,
    LengthReward,
    SequenceSimilarityReward,
    SequenceValidityReward,
)

print("CUDA available:", torch.cuda.is_available())
if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
```

## 4. W&B run configuration

```python
WANDB_PROJECT  = "eshmun-grpo"
WANDB_RUN_NAME = "eshmun-grpo-local-run-1"
OUTPUT_DIR     = "/tmp/eshmun-grpo-output"

# HuggingFace repository to push checkpoints to (must exist or will be created).
# Format: "<username-or-org>/<repo-name>"
HF_REPO_ID = "khairi/eshmun-grpo"
```

## 5. Model & tokenizer

```python
MODEL_ID = "khairi/6L-ALT-f32-student-SFT"

tokenizer = AutoTokenizer.from_pretrained(MODEL_ID)
if tokenizer.pad_token_id is None:
    tokenizer.pad_token_id = tokenizer.eos_token_id

model = AutoModelForCausalLM.from_pretrained(
    MODEL_ID,
    device_map="auto",
    trust_remote_code=True,
)

n_params = sum(p.numel() for p in model.parameters())
print(f"Model: {MODEL_ID}  ({n_params / 1e6:.1f}M parameters)")
```

## 6. Dataset

The dataset must yield tokenized prompts (`input_ids`, `attention_mask`). Columns not in
`{input_ids, attention_mask}` are forwarded as metadata to reward functions via `**kwargs`.

Here `protein_family_stat` is forwarded to `SequenceSimilarityReward` and `LengthReward`.

```python
DATASET_ID = "khairi/swissprot-cot-grpo"
PROMPT_COLUMN = "UserInput"
MAX_PROMPT_LEN = 128

raw = load_dataset(DATASET_ID, split="train").remove_columns(['AssistantAnswer', 'Entry', 'ThinkingText'])

SYSTEM_PROMPT = None  # set to a string to prepend a system message

def _encode(text: str) -> list[int]:
    messages = []
    if SYSTEM_PROMPT:
        messages.append({"role": "system", "content": SYSTEM_PROMPT})
    messages.append({"role": "user", "content": text})
    tokenized = tokenizer.apply_chat_template(
        messages,
        tokenize=True,
        add_generation_prompt=True,
        return_tensors="pt",
        max_length=MAX_PROMPT_LEN,
        truncation=True,
        padding_side="left",
        padding="max_length",
    )

    return tokenized["input_ids"][0]

def tokenize(batch):
    input_ids = [_encode(t) for t in batch[PROMPT_COLUMN]]
    attention_mask = [[1] * len(ids) for ids in input_ids]

    return {"input_ids": input_ids, "attention_mask": attention_mask}

# Keep metadata columns; drop the raw prompt text column only
drop_cols = [PROMPT_COLUMN]
dataset = raw.map(tokenize, batched=True, batch_size=64, remove_columns=drop_cols)
dataset.set_format("torch")

print(dataset)
```

## 7. Reward functions

| Reward | Weight | Purpose |
|---|---|---|
| `SequenceValidityReward` | 0.1 | All characters are valid amino acids (canonical AA) |
| `SequenceSimilarityReward` | 0.6 | Global identity to a family representative |
| `LengthReward` | 0.2 | Length within family min/max range |
| `AminoAcidRepetitionPenalty` | 0.1 | Penalise long repetitive runs |

```python
reward_fns = [
    SequenceValidityReward(),
    SequenceSimilarityReward(non_valid_seq_reward=-1.0),
    LengthReward(min_length=50, max_length=500, out_of_range_penalty=-1.0),
    AminoAcidRepetitionPenalty(min_aa_repetition=10),
]

reward_weights = [0.1, 0.6, 0.2, 0.1]

print("Reward functions:", [type(fn).__name__ for fn in reward_fns])
```

## 8. Trainer with W&B logging and HF Hub push

`GRPOTrainer` emits structured log lines at every `logging_steps` and calls `save_model()`
at every `save_steps`.

- `_WandbHandler` intercepts the log lines and forwards `key=value` metrics to W&B.
- `HubGRPOTrainer` overrides `save_model()` to push both the model and tokenizer to the
  HuggingFace Hub after every local save, tagging each push with the checkpoint step.

```python
class _WandbHandler(logging.Handler):
    """Parses structured trainer log lines and forwards metrics to W&B."""

    _KV_RE = re.compile(r"(\w+)=([\d.e+\-]+)")

    def emit(self, record: logging.LogRecord) -> None:
        msg = record.getMessage()
        metrics: dict[str, float] = {}
        for match in self._KV_RE.finditer(msg):
            key, raw_val = match.group(1), match.group(2)
            if key == "epoch":
                continue
            try:
                metrics[key] = float(raw_val)
            except ValueError:
                pass
        if not metrics:
            return
        step = int(metrics.pop("step", wandb.run.step))
        wandb.log(metrics, step=step)


# Attach to the trainer's logger only — avoids noise from other modules
_trainer_logger = logging.getLogger("eshmun.trainer.grpo.trainer")
_trainer_logger.addHandler(_WandbHandler())


class HubGRPOTrainer(GRPOTrainer):
    """GRPOTrainer that pushes the model and tokenizer to the HF Hub after every save."""

    def __init__(self, *args, hub_repo_id: str, hub_tokenizer, **kwargs):
        super().__init__(*args, **kwargs)
        self._hub_repo_id = hub_repo_id
        self._hub_tokenizer = hub_tokenizer
        self._save_count = 0

    def save_model(self, output_dir: str | None = None) -> None:
        super().save_model(output_dir)          # saves locally as usual

        self._save_count += 1
        commit_message = f"checkpoint-{self._save_count} (step {output_dir})"

        print(f"Pushing to Hub: {self._hub_repo_id} …")
        self.model.push_to_hub(
            self._hub_repo_id,
            commit_message=commit_message,
        )
        self._hub_tokenizer.push_to_hub(
            self._hub_repo_id,
            commit_message=commit_message,
        )
        print(f"  ✓ pushed to https://huggingface.co/{self._hub_repo_id}")
```

## 9. Train

```python
grpo_config = GRPOConfig(
    output_dir=OUTPUT_DIR,
    # --- general ---
    num_epochs=1,
    per_device_train_batch_size=2,
    gradient_accumulation_steps=4,
    learning_rate=1e-5,
    weight_decay=0.01,
    warmup_steps=20,
    max_grad_norm=1.0,
    logging_steps=10,
    save_steps=200,
    # --- GRPO-specific ---
    group_size=4,
    epsilon=0.2,
    beta=0.01,
    num_ppo_epochs=1,
    temperature=1.0,
    max_new_tokens=512,
    reward_weights=reward_weights,
    ref_model_sync_epochs=1,
)

run = wandb.init(
    project=WANDB_PROJECT,
    name=WANDB_RUN_NAME,
    config={
        "model": MODEL_ID,
        "dataset": DATASET_ID,
        "num_epochs": grpo_config.num_epochs,
        "batch_size": grpo_config.per_device_train_batch_size,
        "grad_accum": grpo_config.gradient_accumulation_steps,
        "learning_rate": grpo_config.learning_rate,
        "group_size": grpo_config.group_size,
        "epsilon": grpo_config.epsilon,
        "beta": grpo_config.beta,
        "temperature": grpo_config.temperature,
        "max_new_tokens": grpo_config.max_new_tokens,
        "reward_weights": grpo_config.reward_weights,
        "reward_fns": [type(fn).__name__ for fn in reward_fns],
        "hub_repo_id": HF_REPO_ID,
    },
)

trainer = HubGRPOTrainer(
    model=model,
    config=grpo_config,
    tokenizer=tokenizer,
    reward_fns=reward_fns,
    train_dataset=dataset,
    hub_repo_id=HF_REPO_ID,
    hub_tokenizer=tokenizer,
)

metrics = trainer.train()
print("Training complete:", metrics)
wandb.log({f"final/{k}": v for k, v in metrics.items()})
```

## 10. Save model

```python
trainer.save_model(OUTPUT_DIR)
tokenizer.save_pretrained(OUTPUT_DIR)
print(f"Model saved to {OUTPUT_DIR}")

# Log the checkpoint as a W&B artifact
artifact = wandb.Artifact(name="eshmun-grpo-checkpoint", type="model")
artifact.add_dir(OUTPUT_DIR)
run.log_artifact(artifact)

wandb.finish()
print("W&B run finished.")
```
